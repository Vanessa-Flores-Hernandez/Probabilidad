```python
import matplotlib.pyplot as plt

# Valores de la variable aleatoria
x = [0, 1, 2, 3, 4, 5]
p = [0.05, 0.10, 0.25, 0.30, 0.20, 0.10]

# 1. Verificar suma = 1
print("Suma de probabilidades:", sum(p))

# 2. Graficar distribución de probabilidad
plt.stem(x, p)
plt.title("Distribución de Probabilidad - Ventas Semanales")
plt.xlabel("Unidades vendidas")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()

# 3. Calcular el valor esperado
esperanza = sum(xi * pi for xi, pi in zip(x, p))
print("Valor esperado E(X):", esperanza)

# Calcular E(X²)
esperanza_cuadrado = sum(xi**2 * pi for xi, pi in zip(x, p))
print("E(X²):", esperanza_cuadrado)

# Calcular la varianza
varianza = esperanza_cuadrado - esperanza**2
print("Varianza:", varianza)

# Calcular la desviación estándar
desviacion = varianza**0.5
print("Desviación estándar:", desviacion)

# 4. Calcular la función de distribución acumulada
f = []
acumulada = 0

for pi in p:
    acumulada += pi
    f.append(acumulada)

print("Función de distribución acumulada:", f)

# Graficar función acumulada
plt.step(x, f, where="post")
plt.scatter(x, f)
plt.title("Función de Distribución Acumulada - Ventas")
plt.xlabel("Unidades vendidas")
plt.ylabel("F(x)")
plt.grid(True)
plt.show()

# 5. Interpretación de resultados
print("\nInterpretación de resultados")
print("Número promedio de unidades vendidas:", esperanza)
print("Variabilidad de las ventas:", desviacion)
print("Probabilidad de vender 3 unidades o menos:", f[3])
print("Probabilidad de vender 4 unidades o menos:", f[4])
```
