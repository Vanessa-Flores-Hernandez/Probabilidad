```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr

# 1. Carga de conjuntos de datos
df1 = pd.read_csv("grupo1.csv")
df2 = pd.read_csv("grupo2.csv")
df3 = pd.read_csv("grupo3.csv")

grupos = [("Grupo 1", df1), ("Grupo 2", df2), ("Grupo 3", df3)]

# 2. Procesamiento, ajuste de modelo y visualización por grupo
for nombre, df in grupos:
    print(f"==================== {nombre} ====================")
    print("--- Exploración de Datos ---")
    print(df.describe())

    # Preparación de variables
    X = df[['HorasEstudio']]
    y = df['Calificacion']

    # Entrenamiento del modelo de regresión lineal
    modelo = LinearRegression()
    modelo.fit(X, y)

    m = modelo.coef_[0]
    b = modelo.intercept_

    # Coeficiente de correlación de Pearson y valor p
    r, p_valor = pearsonr(df['HorasEstudio'], df['Calificacion'])
    r2 = r**2

    print("\n--- Resultados del Modelo de Regresión ---")
    print(f"Coeficiente de la pendiente (m): {m:.4f}")
    print(f"Intercepción de la recta (b): {b:.4f}")
    print(f"Ecuación de regresión: Ŷ = {b:.4f} + ({m:.4f}) * X")
    print(f"Coeficiente de correlación de Pearson (r): {r:.4f}")
    print(f"Coeficiente de determinación (r²): {r2:.4f}")
    print(f"Valor p: {p_valor:.4e}\n")

    # Gráfico 1: Histograma de Calificaciones
    plt.figure(figsize=(6, 4))
    sns.histplot(df['Calificacion'], kde=True, color='skyblue')
    plt.title(f'Histograma de Calificaciones - {nombre}')
    plt.xlabel('Calificación Final')
    plt.ylabel('Frecuencia')
    plt.grid(True)
    plt.show()

    # Gráfico 2: Dispersión con Línea de Regresión
    plt.figure(figsize=(6, 4))
    plt.scatter(df['HorasEstudio'], df['Calificacion'], color='blue', label='Datos')
    plt.plot(df['HorasEstudio'], modelo.predict(X), color='red', label='Línea de regresión')
    plt.title(f'Diagrama de Dispersión y Regresión Lineal - {nombre}')
    plt.xlabel('Horas de Estudio (X)')
    plt.ylabel('Calificación Final (Y)')
    plt.legend()
    plt.grid(True)
    plt.show()
```
