```python
from statistics import *
```

```python
#Lista de calificaciones del grupo a de probabilidad y estadistica
grupoa = [70, 70, 65, 90, 100, 45, 80, 73, 20, 100, 96]
print(grupoa)
type(grupoa)
```

```python
# Sacar la medida de las calificaciones con una formula
promedio = sum(grupoa) / len(grupoa)
print(promedio)
```

```python
#utilizar la funcion  mean que esta en la libreria statistics para sacar
#el promedio

promedio2 = mean(grupoa)
print(promedio2)
```

```python
#detreminar el valor de la mediana con la funcion median

mediana = median(grupoa)
print(mediana)
```
