### **Estadísticas descriptivas del precio**

```python
precio = autos["price"].dropna()
```

```python
print("Valor mínimo:", precio.min())
print("Valor máximo:", precio.max())
print("Moda(s):", precio.mode().tolist())
print("Media:", precio.mean())
print("Mediana:", precio.median())
print("Desviación estándar:", precio.std())
```

```python
autos["price"].describe()
```

### **Interpretación**

El precio mínimo de los automóviles es de 5,118, mientras que el precio máximo es de 45,400. La media de los precios es aproximadamente de $13,207.13 y la mediana es de 10,295.

La media es mayor que la mediana, lo que indica que existen algunos automóviles con precios elevados que hacen que el promedio aumente. La desviación estándar es aproximadamente de $7,947.07, por lo que existe una variación considerable entre los precios de los automóviles.

Los resultados de `describe()` permiten observar los cuartiles de los precios. El primer cuartil corresponde a 7,775, la mediana es de 10,295 y el tercer cuartil es de $16,500.

## **2. Histograma de precios de los automóviles**:

### **Número de intervalos utilizando la regla de Sturges**

```python
n = precio.count()
k = 1 + 3.322 * np.log10(n)

print("Número de datos válidos:", n)
print("Número de bins según Sturges:", k)
print("Número de bins utilizado:", int(np.ceil(k)))
```

```python
plt.figure(figsize=(10, 6))

plt.hist(precio, bins=9, alpha=0.7, edgecolor="black")

plt.title("Histograma de precios de los automóviles")
plt.xlabel("Precio del automóvil")
plt.ylabel("Frecuencia")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

### **Interpretación**

En el histograma se puede observar que la mayor cantidad de automóviles se concentra en los rangos de precios bajos y medios. Conforme aumenta el precio, la frecuencia de automóviles disminuye.

También se observa una distribución con una cola hacia la derecha debido a que existen algunos automóviles con precios considerablemente altos. Esto coincide con el hecho de que la media $13,207.13 sea mayor que la mediana 10,295.

### **Conclusión**

La mayoría de los automóviles del dataset tienen precios bajos o medios, mientras que una menor cantidad de automóviles presenta precios elevados. De acuerdo con la regla de Sturges, se utilizaron 9 intervalos para representar la distribución de los precios.

## **3. Diagrama de caja y bigotes: price vs body-style**

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=autos, x="body-style", y="price")

plt.title("Precio de los automóviles según el tipo de vehículo")
plt.xlabel("Tipo de vehículo (body-style)")
plt.ylabel("Precio")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
autos.groupby("body-style")["price"].agg(
    ["count", "min", "median", "mean", "max"]
).round(2)
```

### **Interpretación**

En el diagrama de caja se pueden comparar los precios de los automóviles según su tipo de vehículo.

Los automóviles tipo hatchback presentan generalmente precios más bajos. En cambio, los convertibles y hardtops presentan precios centrales más elevados. Los sedanes tienen una gran cantidad de datos y presentan una variación considerable en sus precios.

También se pueden observar valores atípicos en algunos de los tipos de vehículos, principalmente en los que presentan precios elevados.

### **Conclusión**

El tipo de vehículo presenta diferencias en la distribución de los precios. En este dataset, los hatchbacks tienden a ser más económicos, mientras que los convertibles y hardtops presentan precios más altos.

## **4. Diagrama de caja y bigotes: price vs drive-wheels**

```python
plt.figure(figsize=(9, 6))

sns.boxplot(data=autos, x="drive-wheels", y="price")

plt.title("Precio de los automóviles según el tipo de tracción")
plt.xlabel("Tipo de tracción (drive-wheels)")
plt.ylabel("Precio")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
autos.groupby("drive-wheels")["price"].agg(
    ["count", "min", "median", "mean", "max"]
).round(2)
```

### **Interpretación**

El diagrama de caja muestra diferencias importantes entre los precios de acuerdo con el tipo de tracción.

Los automóviles con tracción rwd presentan los precios centrales más elevados y una mayor dispersión. Los automóviles con tracción fwd presentan precios generalmente más bajos. La categoría 4wd presenta una cantidad menor de automóviles, por lo que sus resultados deben interpretarse con mayor cuidado.

### **Conclusión**

En este dataset, los automóviles con tracción rwd tienden a presentar precios más altos que los automóviles con tracción fwd y 4wd. También presentan una mayor variabilidad en sus precios.

  ## **5. Diagrama de caja y bigotes: price vs engine-location**

```python
plt.figure(figsize=(8, 6))

sns.boxplot(data=autos, x="engine-location", y="price")

plt.title("Precio de los automóviles según la ubicación del motor")
plt.xlabel("Ubicación del motor (engine-location)")
plt.ylabel("Precio")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
autos.groupby("engine-location")["price"].agg(
    ["count", "min", "median", "mean", "max"]
).round(2)
```

### **Interpretación**

La mayoría de los automóviles del dataset tienen el motor ubicado en la parte delantera. Solamente se encuentran algunos automóviles con el motor ubicado en la parte trasera.

Los automóviles con motor trasero presentan precios considerablemente mayores. Sin embargo, este grupo contiene muy pocas observaciones, por lo que no sería adecuado generalizar este resultado a todos los automóviles.

### **Conclusión**

La ubicación del motor muestra una diferencia en los precios dentro del dataset, pero los resultados de los automóviles con motor trasero deben tomarse con precaución debido a la pequeña cantidad de datos disponibles.
