```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Datos
n = 500
media = 70
desviacion = 4

# Calcular valores Z
z1 = (68 - media) / desviacion
z2 = (72 - media) / desviacion

print("Z para 68 kg:", z1)
print("Z para 72 kg:", z2)

# Calcular probabilidades acumuladas
p1 = norm.cdf(68, media, desviacion)
p2 = norm.cdf(72, media, desviacion)

# Calcular probabilidad entre 68 y 72 kg
probabilidad = p2 - p1

print("Probabilidad:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")

# Calcular número de estudiantes
estudiantes = n * probabilidad

print("Número de estudiantes:", round(estudiantes))

# Graficar distribución normal
x = np.linspace(55, 85, 500)
y = norm.pdf(x, media, desviacion)

plt.plot(x, y)
plt.fill_between(
    x, y,
    where=(x >= 68) & (x <= 72),
    alpha=0.5
)

plt.title("Distribución normal del peso de los estudiantes")
plt.xlabel("Peso (kg)")
plt.ylabel("Densidad de probabilidad")
plt.grid(True)
plt.show()

```
