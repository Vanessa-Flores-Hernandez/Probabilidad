**Tipo de distribución: Binomial, con aproximación de Poisson**

La distribución binomial es el modelo exacto porque se conoce el número de motores del embarque (1,000) y la probabilidad de que cada motor sea defectuoso (0.001). Pero, se puede utilizar la distribución de Poisson como aproximación, ya que el número de motores es grande y la probabilidad de defecto es pequeña.

La media se calcula de la siguiente manera:

λ = n × p = 1000 × 0.001 = 1


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Datos
n = 1000
p = 0.001
media = n * p

print("La media es:", media)

# Probabilidad de que ningún motor sea defectuoso
a = poisson.pmf(0, media)

print("La probabilidad de que ningún motor sea defectuoso es:", a)

# Probabilidad de que tres motores sean defectuosos
b = poisson.pmf(3, media)

print("La probabilidad de que tres motores sean defectuosos es:", b)

# Probabilidad de que cuatro motores sean defectuosos
c = poisson.pmf(4, media)

print("La probabilidad de que cuatro motores sean defectuosos es:", c)

# Graficar la distribución de Poisson
x = np.arange(0, 8)
y = poisson.pmf(x, media)

plt.bar(x, y)
plt.title("Distribución de Poisson de motores defectuosos")
plt.xlabel("Número de motores defectuosos")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y")
plt.show()
```
