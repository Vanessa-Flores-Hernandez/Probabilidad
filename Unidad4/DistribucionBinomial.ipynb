```python
from scipy.stats import binom

# parametros de la distribucion binomial
probabilidad_exito = 0.83
num_pacientes = 16
x = 16 # numero de exitos

# calcular la probabilidad de que ningun paciente sufra la enfermedad
probabilidad_ninguno = binom.pmf(x, num_pacientes, probabilidad_exito)

print("La probabilidad de que ningun paciente sufra la enfermedad es: ", probabilidad_ninguno)
```

```python
from scipy.stats import binom

# parametros de la distribucion binomial
probabilidad_exito = 0.83
num_pacientes = 16
x = 0

# calcular la probabilidad de que todos los paciente sufra la enfermedad
probabilidad_todos_enfermos = binom.pmf(x, num_pacientes, probabilidad_exito)

print("La probabilidad de que todos los pacientes sufra la enfermedad es: ", probabilidad_todos_enfermos)
```
