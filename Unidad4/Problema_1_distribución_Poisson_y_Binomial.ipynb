**Tipo de distribución: Poisson**

Se utiliza la distribución de Poisson porque se conoce el promedio de ingresos a cuidados intensivos por día, que es de 5 personas, y se desea calcular la probabilidad de que ocurra una cantidad determinada de ingresos.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Datos
media = 5

# Inciso a: Probabilidad de que ingresen 2 personas
a = poisson.pmf(2, media)

print("La probabilidad de que ingresen 2 personas es:", a)

# Inciso b: Probabilidad de que ingresen más de 10 personas
b = 1 - poisson.cdf(10, media)

print("La probabilidad de que ingresen más de 10 personas es:", b)

# Graficar la distribución de Poisson
x = np.arange(0, 16)
y = poisson.pmf(x, media)

plt.bar(x, y)
plt.title("Distribución de Poisson")
plt.xlabel("Número de personas ingresadas")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y")
plt.show()
```
